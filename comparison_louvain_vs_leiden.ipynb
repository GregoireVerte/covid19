```python
import pandas as pd
import joblib

#### DataFrame 
df = joblib.load('df_with_all_clusters.joblib')
print("DataFrame wczytany.")

#### Tabela Przestawna (Louvain vs Leiden)
comparison_matrix_louvain_leiden = pd.crosstab(
    df['louvain_community'], 
    df['leiden_community']
)

#### zapis do pliku CSV
output_filename = 'comparison_louvain_vs_leiden.csv'
comparison_matrix_louvain_leiden.to_csv(output_filename)

print(f"\nPełna tabela porównawcza (Louvain vs Leiden) została zapisana do pliku: {output_filename}")
```

```python
## Heatmap Louvain vs Leiden

import seaborn as sns
import matplotlib.pyplot as plt


csv_filename = 'comparison_louvain_vs_leiden.csv'
comparison_matrix = pd.read_csv(csv_filename, index_col=0) ### index_col=0 --> pierwsza kolumna jako indeks (ID społeczności Louvain)
print("Dane wczytane.")

plt.figure(figsize=(18, 16))

### logarytmiczna skala kolorów (norm=LogNorm()) aby lepiej uwidocznić mniejsze wartości
from matplotlib.colors import LogNorm
sns.heatmap(comparison_matrix, cmap="viridis", norm=LogNorm())

plt.title('Mapa Zgodności: Społeczności Louvain vs Leiden', fontsize=16)
plt.xlabel('ID Społeczności Leiden')
plt.ylabel('ID Społeczności Louvain')
plt.show()
```
